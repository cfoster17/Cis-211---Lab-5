# L7 – Dictionaries

A list keeps items in numbered positions, which is perfect when the *order* matters. But much real-world data is naturally organized by **name**, not by position: look up a price by the *product's name*, a phone number by a *person's name*, a definition by a *word*.

Python's **dictionary** stores **key-value pairs**: you provide a key, it hands back the matching value. You met this exact idea back in Week 1 – DNS is a giant dictionary whose keys are domain names and whose values are IP addresses – and the contacts app on your phone is another one: names in, numbers out.

# Creating a Dictionary

A dictionary is written in curly braces, as `key: value` pairs separated by commas. L1 showed you one briefly when exploring data types; here is a more useful one – a bakery's menu, where each product name (the key) maps to its price (the value).

````python
menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75}
empty = {}

print(menu)
print(type(menu))
print(len(menu))        # len counts the PAIRS
````

# Looking Up a Value by Its Key

Where a list uses an index in square brackets, a dictionary uses a **key** in square brackets. Keys are case-sensitive, and looking up a key that does not exist raises a `KeyError`.

````python
menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75}

print(menu['cupcake'])
print(menu['muffin'])
````

## The Safe Lookup: `get()`

`menu.get(key)` does the same lookup, but a missing key gives you `None` instead of crashing – and you can supply your own fallback value as a second argument. Use `get()` whenever the key might not be there.

````python
menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75}

print(menu.get('cookie'))                       # found: 2.0
print(menu.get('bagel'))                        # missing: None (no crash)
print(menu.get('bagel', 'not on the menu'))     # missing: your fallback
````

# Adding, Updating, and Removing Pairs

Dictionaries are mutable, like lists. Assigning to a key either **adds** the pair (new key) or **updates** it (existing key) – same syntax for both.

````python
menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75}
print('Starting menu:', menu)

menu['scone'] = 3.25            # NEW key: adds a pair
print('After adding: ', menu)

menu['cookie'] = 2.25           # EXISTING key: updates the value
print('After update: ', menu)

del menu['muffin']              # removes a pair by key
print('After delete: ', menu)

price = menu.pop('scone')       # removes a pair AND hands back its value
print('After pop:    ', menu, '| popped value:', price)
````

# Checking for a Key: the `in` Operator

Just like lists – but note that `in` checks the **keys**, not the values. Combined with L3's if-else, this is how you avoid KeyErrors.

````python
menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75}

print('cookie' in menu)     # True - it is a key
print(3.50 in menu)         # False - 3.50 is a VALUE, not a key

item = 'muffin'
if item in menu:
    print(f'{item} costs ${menu[item]:.2f}')
else:
    print(f'Sorry, we do not sell {item}.')
````

# Looping Over a Dictionary

A `for` loop over a dictionary visits its **keys**. From each key you can reach its value. The `.items()` method hands you both at once, which is usually the cleanest.

Three related methods: `.keys()` (all keys), `.values()` (all values), `.items()` (all pairs).

````python
menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75, 'scone': 3.25}

# looping over keys (the default)
for item in menu:
    print(f'{item} costs ${menu[item]:.2f}')
````

````python
menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75, 'scone': 3.25}

# looping over pairs with .items() - note the TWO loop variables
for item, price in menu.items():
    print(f'{item} costs ${price:.2f}')
````

````python
menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75, 'scone': 3.25}

print(list(menu.keys()))
print(list(menu.values()))
print(f'Average price: ${sum(menu.values()) / len(menu):.2f}')
````

# Counting with a Dictionary

A classic and powerful pattern: counting how many times each thing appears. Walk through the data; for each item, add 1 to its count, creating the count at 0 the first time you see it. Businesses use exactly this to answer “what sold most today?”

````python
todays_orders = ['cupcake', 'cookie', 'cupcake', 'muffin',
                 'cupcake', 'cookie', 'scone', 'cupcake']

counts = {}                             # empty dictionary as the accumulator
for item in todays_orders:
    counts[item] = counts.get(item, 0) + 1    # old count (or 0) plus one

print(counts)
````

# Dictionaries and Functions Together

Just as a function can take a list (L6), it can take a dictionary. Here is everything in this notebook working as one small system: a lookup function with a safe default, and an order-totaling function that combines a dictionary (the menu) with a list (the order).

````python
def lookup_price(menu, item):
    return menu.get(item, 0)        # unknown items cost 0 and can be flagged

def order_total(menu, order):
    total = 0
    for item in order:              # the accumulator pattern from L6
        total = total + lookup_price(menu, item)
    return total

menu = {'cupcake': 3.50, 'cookie': 2.00, 'muffin': 2.75, 'scone': 3.25}
order = ['cupcake', 'cupcake', 'cookie', 'scone']

print(f'Order total: ${order_total(menu, order):.2f}')
````
