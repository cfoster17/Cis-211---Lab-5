# L6 – Lists

Every variable you have used so far holds exactly **one** value. But real data comes in groups: a store's menu, a cart full of items, a month of daily sales. A **list** is Python's basic container for an ordered collection of values, and it is the data structure you will use more than any other.

You have already *met* lists briefly: in L1 (`["apple", "banana", "cherry"]` when exploring data types) and in L4 (looping over `[10, 20, 34, 50, 55]`). Now we study them properly.

# Creating a List

A list is written as values separated by commas, inside square brackets. A list can hold any type – strings, numbers, Booleans – and even a mix.

```python
menu = ['cupcake', 'cookie', 'muffin', 'scone', 'tart']
prices = [3.50, 2.00, 2.75, 3.25, 4.00]
mixed = ['cupcake', 3.50, True]
empty = []

print(menu)
print(prices)
print(type(menu))
```

```python
# len() tells you how many items a list holds
print(len(menu))
print(len(empty))
```

# Indexing: Grabbing One Item

Each item has a numbered position called its **index** – and indexes start at **0**, not 1. In a five-item list, the valid indexes are 0, 1, 2, 3, 4.

Python also supports **negative indexes**, which count from the end: -1 is the last item, -2 the second-to-last, and so on. That is the clean way to say “the last one” without knowing the list's length.

```python
menu = ['cupcake', 'cookie', 'muffin', 'scone', 'tart']
#          0          1          2         3        4
#         -5         -4         -3        -2       -1

print(menu[0])      # first item
print(menu[2])      # third item
print(menu[-1])     # last item
print(menu[-2])     # second-to-last item
```

# Slicing: Grabbing a Range of Items

A **slice** copies a piece of a list: `my_list[start:stop]` gives the items from index `start` up to **but not including** `stop`. Sound familiar? It is the exact same “stop is never included” rule as `range()` in L4.

```python
menu = ['cupcake', 'cookie', 'muffin', 'scone', 'tart']

print(menu[1:3])    # indexes 1 and 2 (3 is NOT included)
print(menu[:2])     # from the beginning up to (not including) index 2
print(menu[2:])     # from index 2 to the end
print(menu[:])      # the whole list (a copy)
```

# Lists Are Mutable: You Can Change Them

Unlike a number or a string, a list can be modified after it is created – programmers say lists are **mutable**. You can replace an item by assigning to its index, and you can grow or shrink the list with its built-in methods:

| Method / statement | What it does |
|---|---|
| `my_list[i] = value` | Replace the item at index i |
| `my_list.append(value)` | Add a value to the **end** |
| `my_list.insert(i, value)` | Insert a value at index i, shifting the rest right |
| `my_list.remove(value)` | Remove the **first** occurrence of a value |
| `my_list.pop(i)` | Remove (and give back) the item at index i; no i means the last item |
| `del my_list[i]` | Delete the item at index i |

```python
menu = ['cupcake', 'cookie', 'muffin']
print('Starting menu:   ', menu)

menu[1] = 'oatmeal cookie'          # replace by index
print('After replace:   ', menu)

menu.append('scone')                # add to the end
print('After append:    ', menu)

menu.insert(0, 'brownie')           # insert at the front
print('After insert:    ', menu)

menu.remove('muffin')               # remove by value
print('After remove:    ', menu)

last = menu.pop()                   # remove the last item (and keep it)
print('After pop:       ', menu, '| popped:', last)
```

# Checking Membership: the `in` Operator

`value in my_list` is a Boolean expression (True or False), so it plugs directly into the `if` statements you learned in L3.

```python
menu = ['cupcake', 'cookie', 'muffin', 'scone']

print('cookie' in menu)
print('bagel' in menu)

if 'scone' in menu:
    print('Yes, we sell scones!')
```

# Looping Over a List

You saw this in L4; it is worth naming the two styles.

- **By item** – the common case: you want each value.
- **By index** – when you also need the position, use `range(len(my_list))`.

```python
prices = [3.50, 2.00, 2.75, 3.25]

# style 1: by item
for p in prices:
    print(f'${p:.2f}')
```

```python
menu = ['cupcake', 'cookie', 'muffin', 'scone']

# style 2: by index - useful when the position matters
for i in range(len(menu)):
    print(f'{i + 1}. {menu[i]}')
```

# The Accumulator Pattern

The most important loop-plus-list pattern in business programming: start a running total at 0, then add each item to it as the loop visits it. This is literally how a shopping-cart total works.

```python
prices = [3.50, 2.00, 2.75, 3.25]

total = 0                   # 1. start the accumulator at zero
for p in prices:            # 2. visit every price
    total = total + p       # 3. add it to the running total

print(f'Cart total: ${total:.2f}')
```

```python
# Python has shortcuts for the most common list math:
prices = [3.50, 2.00, 2.75, 3.25]

print(sum(prices))              # total
print(min(prices))              # cheapest
print(max(prices))              # most expensive
print(sum(prices) / len(prices))    # average
```

# Sorting and Combining Lists

```python
prices = [3.50, 2.00, 2.75, 3.25]

print(sorted(prices))               # sorted() gives a NEW sorted list
print(prices)                       # ...the original is untouched

prices.sort()                       # .sort() changes the list itself
print(prices)

prices.sort(reverse=True)           # highest to lowest
print(prices)
```

```python
# + joins lists; * repeats them
drinks = ['coffee', 'tea']
food = ['cupcake', 'muffin']

print(drinks + food)
print(drinks * 3)
```

# Lists and Functions Together

L5 plus L6: a function that takes a whole list as its parameter and returns one answer. This pattern – *data goes in, a summary comes out* – is the heart of data analysis.

```python
def average(numbers):
    return sum(numbers) / len(numbers)

daily_sales = [412.50, 389.25, 501.75, 444.00, 467.80]
print(f'Average daily sales: ${average(daily_sales):.2f}')
```
