# L5 – Functions

So far, every program you have written runs as one single block of code, top to bottom. That works for small programs, but real programs – a checkout page, a web app, a game – are built from many small, named, reusable pieces called **functions**.

You have already been *using* functions all semester: `print()`, `input()`, `int()`, `type()`, and `range()` are all functions that someone else wrote. In this notebook you learn to write your own.

# Why Functions?

Imagine a store's checkout needs to calculate sales tax in three different places: the cart page, the checkout page, and the emailed receipt. Without functions you would copy the same tax code three times – and when the tax rate changes, you must remember to fix it in all three places. Miss one, and the receipt disagrees with the cart.

With a function, the tax calculation is written **once**, given a name, and *called* from all three places. One fix updates everything.

Functions give you:

- **Reuse** – write once, call many times
- **Organization** – a big problem becomes small named pieces (programmers call this *divide and conquer*)
- **Readability** – `calculate_tax(total)` says what it does; forty lines of raw math does not
- **Easier debugging** – if tax is wrong, you know exactly which piece to inspect

# Defining and Calling a Function

A function definition has two parts:

- A **header**: the keyword `def`, the function's name, parentheses, and a colon
- A **body**: one or more indented statements (the same indentation rule as `if` and loops)

Function names follow the same rules as variable names: start with a letter or underscore, no spaces, case-sensitive.

```python
# Defining a function (nothing happens yet - Python just memorizes it)
def show_welcome():
    print('Welcome to the Bright Bakery online store!')
    print('Open 24 hours a day, 7 days a week.')
```

```python
# Calling the function makes its body run
show_welcome()
```

```python
# Once defined, a function can be called as many times as you like
show_welcome()
show_welcome()
show_welcome()
```

## Important: Defining Is Not Running

Running the `def` cell only *teaches* Python the function. The body does not execute until you **call** the function by name with parentheses.

Also note the order: a function must be defined **before** it is called. If you call a function Python has not seen yet, you get a `NameError`. In a notebook, that means: always run the cell with the `def` before the cell with the call.

# Parameters: Sending Information Into a Function

The function above always prints the same thing. Functions become truly useful when you can pass them information. A **parameter** is a variable named in the function header; an **argument** is the actual value you send when calling.


```python
def greet(name):            # name is a PARAMETER
    print(f'Hello, {name}! Welcome back.')

greet('Jordan')             # 'Jordan' is an ARGUMENT
greet('Priya')
greet('Chris')
```

```python
# A function can have several parameters, separated by commas.
# Arguments are matched to parameters by position: first to first, second to second.

def show_item(item, price):
    print(f'{item} costs ${price:.2f}')

show_item('cupcake', 3.5)
show_item('cookie', 2)
```

## Keyword Arguments and Default Values

Two conveniences you will see often:

- **Keyword arguments** name the parameter explicitly when calling, so order no longer matters.
- **Default values** in the header make a parameter optional.

```python
def show_item(item, price):
    print(f'{item} costs ${price:.2f}')

# keyword arguments: order does not matter when you name them
show_item(price=2.75, item='muffin')
```

```python
def brew_coffee(size='medium'):     # 'medium' is the DEFAULT value
    print(f'Brewing one {size} coffee...')

brew_coffee('large')    # uses the argument you pass
brew_coffee()           # no argument? the default kicks in
```

# Local Variables and Scope

A variable created inside a function (including its parameters) is **local**: it exists only while the function runs, and it is invisible outside. This is a feature, not a limitation – it means functions cannot accidentally trample each other's variables.

```python
def calculate_tax(amount):
    tax = amount * 0.08875      # tax is LOCAL to this function
    print(f'Tax on ${amount:.2f} is ${tax:.2f}')

calculate_tax(100)

# Uncomment the next line and run the cell: you get a NameError,
# because tax only exists inside the function.

# print(tax)
```

# Returning Values

`print()` *shows* a result on the screen, but the calling code cannot use it. To send a result **back** to the caller – so it can be stored, compared, or used in more math – a function uses the `return` statement.

This is the single most important idea in this notebook.

```python
def rectangle_area(length, width):
    area = length * width
    return area                 # send the result BACK to the caller

# The returned value can be stored in a variable...
room = rectangle_area(12, 4)
print(f'The area is {room} square cm')

# ...or used directly inside other expressions
print(rectangle_area(10, 5) + rectangle_area(8, 4))
```

```python
# return ends the function immediately - any code after it is skipped.
# Combined with if/else (from L3), this lets a function make decisions:

def shipping_fee(order_total):
    if order_total >= 35:
        return 0.0
    else:
        return 5.99

print(shipping_fee(50))     # big order: free shipping
print(shipping_fee(20))     # small order: pays the fee
```

```python
# Functions can return True or False - perfect for yes/no questions.
# By convention, such functions get names that read like questions.

def is_even(number):
    if number % 2 == 0:
        return True
    else:
        return False

print(is_even(10))
print(is_even(7))

# A returned Boolean plugs straight into an if statement:
if is_even(42):
    print('42 is even!')
```

```python
# A function with no return statement returns the special value None.

def say_hi():
    print('hi')

result = say_hi()
print(result)          # None - say_hi PRINTS but does not RETURN
```

## print vs. return – Get This Straight Now

| | What it does | Who sees the value |
|---|---|---|
| `print(x)` | Displays x on the screen | A human reading the output |
| `return x` | Sends x back to the calling code | The rest of your program |

If you need the result for *further work* – adding it, comparing it, saving it – you need `return`. A function that only prints is a dead end: its result cannot be reused.

# Functions Working Together

Real programs chain functions: one function's return value becomes another's argument, and loops call functions over and over. Everything from L3 and L4 combines here.

```python
def fahrenheit_to_celsius(f):
    return (f - 32) / 1.8

# A loop (L4) calling a function (L5) and printing a table:
print('Fahrenheit \t Celsius')
print('-----------------------')
for f in range(50, 101, 10):
    c = fahrenheit_to_celsius(f)
    print(f'{f}\t\t {c:.1f}')
```

```python
# Putting the whole module together: a tiny checkout.

def shipping_fee(order_total):
    if order_total >= 35:
        return 0.0
    return 5.99            # note: no else needed - if we got here, the if was False

def final_total(order_total):
    return order_total + shipping_fee(order_total)

print(f'${final_total(20.00):.2f}')    # 25.99
print(f'${final_total(60.00):.2f}')    # 60.00
```
